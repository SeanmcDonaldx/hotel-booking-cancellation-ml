### Initial Data Audit Summary

- The dataset contains 119,390 bookings and 32 columns.
- The target variable is `is_canceled`. Around 63% of bookings were not cancelled and 37% were cancelled.
- Missing values were found in `children`, `country`, `agent` and `company`, with `company` having the most missing values.
- There are 31,994 exact duplicate rows.
- 180 bookings have no adults, children or babies recorded.
- Some unusual numerical values were found, including an ADR of -6.38 and an extreme ADR of 5400.
- `reservation_status` directly reveals whether a booking was cancelled, while `reservation_status_date` contains information about the final booking outcome. Both would cause data leakage.
- `assigned_room_type`, `booking_changes` and `days_in_waiting_list` may contain information that is only known after the booking is initially made, so these will also not be used as predictors.
- Some large guest counts and other extreme values will be investigated before deciding whether they should be removed.
- The prediction is intended to be made using information available when the booking is initially made. Features that are only known after this point should not be used because they could cause data leakage.

### Data Cleaning

- `reservation_status` and `reservation_status_date` were removed because they contain information about the final booking outcome. `assigned_room_type`, `booking_changes` and `days_in_waiting_list` were also removed because they may contain information that is not available when the booking is initially made.
- Removed `company` because around 94% of its values were missing.
- The 4 missing children values are handled using median imputation in the preprocessing pipeline.
- Missing `country` values were labelled as `Unknown`.
- Missing `agent` values were replaced with 0 to represent no recorded agent.
- Removed 180 records where the total number of guests was zero.
- 32,246 identical rows were found but were not removed because the dataset does not contain a unique booking identifier, so they may represent separate bookings.
- After cleaning, the dataset contains 119,210 rows and 26 columns.
- After cleaning, 62.92% of bookings were not cancelled and 37.08% were cancelled.

### Preprocessing Pipeline

A `ColumnTransformer` was used to apply different preprocessing to numerical and categorical features. Missing numerical values are filled using median imputation and numerical features are then standardised using `StandardScaler`. Categorical features are encoded using `OneHotEncoder(handle_unknown="ignore")`.

The preprocessing and baseline classifier were combined in a `Pipeline`. This means the preprocessing is fitted using the training data and then applied consistently to the test data, helping to avoid data leakage.
